```python
df = pd.read_csv("/Users/zgl12/Modules/DES_Param_Analysis/run_publication_registry.csv")

baseline_val = df[df['run_name'] == 'baseline']["logZ"].values[0]
```

```python
def section_df(df, substring):
    temp_df = df[df['run_name'].str.contains(substring)]
    return temp_df[temp_df['logZ'] >= (baseline_val+1)]
    
    
```
